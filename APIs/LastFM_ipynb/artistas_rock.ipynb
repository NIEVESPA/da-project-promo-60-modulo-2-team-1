```python
url_api = "http://ws.audioscrobbler.com/2.0/"
```

### ------------------------------- Hacemos una función para llamar a la api ------------------------------- 

```python
def llamar_lastfm_api(method, artist_name, cliente):

        params = {'method': method,'artist': artist_name,'api_key': cliente,'format': 'json'}

        response = requests.get(url_api, params=params)


        if response.status_code == 200:
            return response.json()
        else:
            print(f"Error {response.status_code}: No se pudo obtener la información para el método {method}")
            return
```

### ------------------------------- Hacemos un bucle for para iterar por cada artista -- en esta prueba lo hacemos con prints de control porque son solo cuatro artistas ------------------------------- 

```python
artistas_a_consultar_prueba = ["Radiohead", "Pink Floyd", "Led Zeppelin", "Queen",  "The Beatles"]
lista_artistas_prueba = []


for nombre_artista in artistas_a_consultar_prueba:
    print(f"--- Consultando: {nombre_artista} ---")
    
    datos_artista = llamar_lastfm_api('artist.getinfo', nombre_artista, cliente)  
    
    
    if datos_artista and 'artist' in datos_artista:
        
        artista_data = datos_artista['artist']
        
        
        oyentes = artista_data.get('stats', {}).get('listeners', 'arista no encontrado') 
        reproducciones = artista_data.get('stats', {}).get('playcount', 'N/D')
        
        bio_content = artista_data.get('bio', {}).get('content', '')
        
        if not bio_content or bio_content.strip() == "":
            bio_resumen = 'biografia no encontrada'
        else:
            bio_resumen = artista_data.get('bio', {}).get('summary')
        
        lista_artistas_dict = {'artista': artista_data.get('name', nombre_artista),'oyentes': oyentes, 'reproducciones': reproducciones, 'bio_resumen': bio_resumen}
        lista_artistas_prueba.append(lista_artistas_dict)
        
        print(f"Datos de {nombre_artista} procesados y guardados.")
           
    else:
        print(f"No se pudo obtener información de {nombre_artista}")


```

### ----Extraemos el país de la biografía por que la api lastfm no nos facilita ese dato-------

```python
# --- 1. DICCIONARIO BASE (PAÍS: GENTILICIO) ---
# Usamos el diccionario que generamos previamente:

PAIS_A_GENTILICIO = {
    "Afghanistan": "Afghan",
    "Albania": "Albanian",
    "Algeria": "Algerian",
    "Argentina": "Argentinian",
    "Australia": "Australian",
    "Austria": "Austrian",
    "Belgium": "Belgian",
    "Bolivia": "Bolivian",
    "Brazil": "Brazilian",
    "Canada": "Canadian",
    "Chile": "Chilean",
    "China": "Chinese",
    "Colombia": "Colombian",
    "Costa Rica": "Costa Rican",
    "Cuba": "Cuban",
    "Denmark": "Dane",
    "Egypt": "Egyptian",
    "France": "French",
    "Germany": "German",
    "Greece": "Greek",
    "India": "Indian",
    "Indonesia": "Indonesian",
    "Ireland": "Irish",
    "Italy": "Italian",
    "Jamaica": "Jamaican",
    "Japan": "Japanese",
    "Mexico": "Mexican",
    "Netherlands": "Dutch",
    "New Zealand": "New Zealander",
    "Nigeria": "Nigerian",
    "Norway": "Norwegian",
    "Peru": "Peruvian",
    "Poland": "Pole",
    "Portugal": "Portuguese",
    "Russia": "Russian",
    "South Korea": "South Korean",
    "Spain": "Spaniard",
    "Sweden": "Swede",
    "Turkey": "Turk",
    "Ukraine": "Ukrainian",
    "United Kingdom": "Briton",
    "United States": "American",
    "United States_adj": "American", 
    "United Kingdom_adj": "British",
    "France_adj": "French",
    "Germany_adj": "German",
    "Spain_adj": "Spanish"}

# --- 2. DICCIONARIO INVERTIDO PARA BÚSQUEDA EFICIENTE ---
# Clave: Gentilicio (lo que busco en la bio), Valor: País
GENTILICIO_A_PAIS = {gentilicio.lower(): pais.replace('_adj', '') 
    for pais, gentilicio in PAIS_A_GENTILICIO.items()}
```

```python
def extraer_pais_de_biografia(bio_resumen, gentilicio_a_pais_map):
  
    bio_normalizada = bio_resumen.lower()


    for gentilicio, pais in gentilicio_a_pais_map.items():

        # Usamos ' ' + gentilicio + ' ' para buscar la palabra completa y evitar coincidencias parciales
        if f' {gentilicio} ' in bio_normalizada or bio_normalizada.startswith(f'{gentilicio} '):
            return pais
        
        # Caso especial para "is an American rock band" (no tiene espacio antes)
        if bio_normalizada.endswith(gentilicio): 
             return pais

    return ''
```

#### Agregamos la key "pais" a nuestra lista

```python
for artista_data in lista_artistas:
  
    bio = artista_data['bio_resumen']
    pais_encontrado = extraer_pais_de_biografia(bio, GENTILICIO_A_PAIS)
    
    artista_data['pais'] = pais_encontrado #aqui agregamos la clave pais en nuestro diccionario
```
