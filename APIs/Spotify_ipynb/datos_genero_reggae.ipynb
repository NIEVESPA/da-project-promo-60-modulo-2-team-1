```python
listacanciones_reggae = []
for anio in range (2000, 2005):
    for offset in range (0,150,50):
     
     datos = sp.search(q=f"genre:reggae, year:{anio}", type="track", limit=50, offset=offset)

     for dato in datos['tracks']['items']:
         
         artista = dato['artists'][0]['name']
         genero = 'latin pop'
         nombre = dato['name']
         tipo = dato['type']
         anio = dato['album']['release_date'][:4]
         spotify_id = dato['id']
       
         informacion_artista = {'Artista': artista, 'Género': 'reggae', 'Tipo': tipo, 'Nombre': nombre, 'Año': anio,'ID': spotify_id} 

         listacanciones_reggae.append(informacion_artista)

print(listacanciones_reggae)     

```

```python
#CREACIÓN DEL DATAFRAME
df_reggae = pd.DataFrame(listacanciones_reggae)

df_reggae
```
