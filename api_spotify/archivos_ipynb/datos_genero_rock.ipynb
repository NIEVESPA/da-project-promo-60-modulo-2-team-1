```python
lista_rock = []

for año in range(2000,2005):
    for offset in range(0, 150, 50):
        
        datos = sp.search(q=f"genre:rock year:{año}", type="track", limit=50, offset=offset)
        
        for dato in datos['tracks']['items']:
            
            artista = dato['artists'][0]['name']
            nombre = dato['name']
            tipo = dato['type']
            fecha_lanzamiento = dato['album']['release_date'][:4]
            spotify_id = dato['id']
            
          
            lista_rock.append({'Artista': artista,'Género': 'rock', 'Tipo': tipo,'Nombre': nombre,'Año_Lanzamiento': fecha_lanzamiento,'ID_Spotify': spotify_id})

```

```python
import pandas as pd
```

```python
df_rock = pd.DataFrame(lista_rock)
df_rock
```

```python
df_rock.to_csv("lista_rock.csv")
```
