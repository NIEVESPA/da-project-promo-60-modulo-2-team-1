```python
## convertimos en tupla lista artista para poder iterar por ella y pasarla a lastfm
```

```python
def convertir_df(ruta):
    df = pd.read_csv(ruta)
    lista_listas = df.values.tolist()

    lista_tuplas = []
    for lista in lista_listas:
        t = tuple(lista)
        lista_tuplas.append(t)

    return lista_tuplas
```

```python
df_latin_pop_tuplas = convertir_df("df_latin_pop.csv")
```

```python
df_latin_pop_tuplas
```

```python
## saco de mi lista de tuplas una lista de artistas para traspasarla a lastfm
```

```python
def artista_tupla(lista_tuplas):
    lista_artista = []
    for dato in lista_tuplas:
        artista = dato[0]
        lista_artista.append(artista)
    print(lista_artista)
    return lista_artista
```

```python

artista_tupla(df_latin_pop_tuplas)
```

```python
lista_artista_latin_pop = artista_tupla(df_latin_pop_tuplas)
```

```python
type(lista_artista_latin_pop)
```

```python
nombre_archivo = 'artistas_latin_pop.json' # Nombre del archivo donde se guardará
```

```python
lista_artista_latin_pop

with open(nombre_archivo, 'w', encoding='utf-8') as archivo:
    
    json.dump(lista_artista_latin_pop, archivo, indent=4) # Usamos json.dump para escribir la lista en el archivo, indent=4 es la identacion en un archivo json

print(f"Lista de artistas guardada en {'Artista_latin_pop.json'}")
```
