```python
auth_manager = SpotifyClientCredentials(cliente, id_secreto)
sp = spotipy.Spotify(auth_manager=auth_manager)
```

```python
lista_electronica = [] #lista vacia antes del bucle

for ano in range(2000, 2005):
    for offset in range (0,150,50):
        
        query = sp.search(q=f"genre:electronic, year:{ano}", type="track", limit=50, offset=offset)

        for dato in query['tracks']['items']:
            artista = dato['artists'][0]['name']
            genero = 'latin pop'
            nombre = dato['name']
            tipo = dato['type']
            ano = dato['album']['release_date'][:4]
            spotify_id = dato['id']

            informacion_artista= {'Artista': artista, 'Género': 'electronic', 'Tipo': tipo, 'Nombre': nombre, 'Año': ano,'ID': spotify_id} #creamos un diccionario con la info que queremos recopila de cada truck. nota: genero str

            lista_electronica.append(informacion_artista)

print(lista_electronica)
```
